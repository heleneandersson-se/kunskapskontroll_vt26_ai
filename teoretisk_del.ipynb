5. RMSE är utvärderingsmått som används för att uppskatta hur väl en regressionsmodell preserar. Det står för Root Mean Square Error. Error är residualen för en prediktion, dvs skillnaden mellan det predikterade värdet och det sanna värdet. Squared innebär att residualen kvadreras. Detta görs för att positiva och negativa skillnader annars kan ta ut varanda. Mean innebär att man sedan tar medelvärdet av residualerna. Slutligen har vi Root, vilket innebär att vi tar ut kvardatroten av värdet. Detta gör det mer läsbart eftersom vi tidigare i algoritmen har kvadrerat värdena. Med RMSE vill man ha ett så lågt värde som möjligt eftersom det i princip är ett medelavstånd mellan sanna och predikterade värden.

7. Confusion Matrix är ett sätt att utvärdera klassifikationsproblem. Det är en matris med alla klasser där en riktning representerar de predikterade värdena och den andra de sanna värdena. Målet är att få höga siffror i diagonalen där sanna värden och predikterade värden sammanfaller. En Confusion Matrix kommer också att visa vilka klasser modellen ofta uppskattar fel.

8. Precision är ett utvärderingsmått för klassifikationsproblem som mäter andelen av de positiva observationerna som är korrekta (True Positive / (True Positive + False Positive)). Precision kommer vara hög om modellen bara uppskattar en observation som positiv när den är väldigt säker, men säger inget om huruvida den missar många positiva fall. Recall blir då låg.

9. Recall mäter andelen av den positiva klassen som predikterats korrekt (True Positive / (True Positive / False Negative)). Hög recall innebär att den positiva klassen sällan uppskattas fel, men det kan också finnas många observationer som felaktiskt uppskattas som postiva (False Positive). Precision blir då låg, så det är en tradeoff. Vad som prioriteras beror på problemet.

11. Hyperparametrar styr hur modellen beter sig medan parametrar är resultatet från modellen (vad vi lärt oss). Ett exempel på en hyperparameter är regulariseringsgraden (C) på t.ex Logistic Regression. Ett exempel på en parameter kan vara interceptet i en linjär modell, förutsatt att dte är något modellen lär sig under träningen. Man kan inte optimera en parameter inför träningen eftersom det är resultatet av träningen. Däremot kan man optimera hyperparametrar för att få så bra resultat av träningen som möjligt.

12. Overfitting betyder att träningen har anpassat sig för mycket till träningsdatan, och därför är dålig på att tolka ny data. Underfitting är tvärt om att modellen inte har lärt sig tillräckligt mycket av datan. Med overfitting kommer modellen prestera bra på träningsdata men dåligt på validering- och testdata medan underfitting presterar dåligt överag. Båda kan åtgärdas genom att justera hyperparametrar eller byta modell. I vissa fall (i synnerhet med underfitting) kan problemet vara att datan inte räcker till för att hitta relevanta mönster.

14. Regularisering för att en modell följer träningsdatan mindre precist. Det begänsar den, vilket gör den meindre känslig för fluktuationer i träningsdatan. Det kan leda till högre bias (mer fel) men lägre varians. Om en modell har för hög varians så kan det leda till overfitting, vilket gör att den predikterar dåligt på ny data. Regularisering är därför ett sätt att göra modellen bättre på att generalisera genom att göra den mindre specifikt anpassad till träningsdatan. 

15. Bias beror på felaktiga modellantaganden och hög bias kan leda till underfitting.
Varians beror på att modellen är känslig för fluktationer i träningsdatan när den tränas och kan leda till overfitting.
Bias och Varians tenderar att vara motpoler till varandra så att högre bias leder till lägre varians och vice versa. Man kan därför prata om en bias/variance tradeoff, där målet är att ha en modell som är balanserad mellan dem.   

16. När man ökar en modells komplexitet ökar man genrellt dess varians men miskar dess bias. Det kan leda till att modellen blir overfitted. Eftersom komplexa modeller kan vara känsliga för fluktationer i datan (hög varians) kan de behöva tränas på mer data.

Kombinationerna kan därefter utvärderas mot varanda med ett utvärderingsmått, t.ex MSE för regressionsproblem.
